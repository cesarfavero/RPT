```python
import gc
import re
import math
import json
import shutil
from pathlib import Path

import torch
import torch.nn as nn
from datasets import Dataset
from peft import LoraConfig, get_peft_model, AutoPeftModelForCausalLM
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    DataCollatorForLanguageModeling,
    Trainer,
    TrainingArguments,
)

HF_TOKEN = os.environ.get('HF_TOKEN')
BASE_MODEL_NAME = 'google/gemma-2-2b-it'
PRUNED_CHECKPOINT = CHECKPOINTS_DIR / 'student_pruned'
PROMPT_TEMPLATE = 'Responda de forma curta e direta.\nPergunta: {question}\nResposta:'
MAX_NEW_TOKENS = 24
BATCH_SIZE = 1
# Jump schedule after l21-adjust: 24 -> 36 -> 48 -> 60 -> 78.
# If a jump drops below target, bisect between the last good point and the failed point.
TARGET_LAYER_LIMITS = [30]
PILOT_LAYER_NAME_FILTERS = ['mlp']
TRAIN_STEPS = 150
LEARNING_RATE = 6e-5
LORA_R = 24
LORA_ALPHA = 48
LORA_DROPOUT = 0.05
TARGET_ACCURACY = 0.92
REFERENCE_RECOVERY_ACCURACY = 0.91
MIN_ACCEPTABLE_ACCURACY = 0.90

RECOVERY_BASELINE = REFERENCE_RECOVERY_ACCURACY

print('Using pruned checkpoint:', PRUNED_CHECKPOINT)
print('Reference recovery baseline:', RECOVERY_BASELINE)
print('Target accuracy:', TARGET_ACCURACY)
print('Minimum acceptable accuracy:', MIN_ACCEPTABLE_ACCURACY)


def cleanup_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


def parse_answer(full_text):
    text = full_text.split('Resposta:')[-1].strip()
    text = re.sub(r'^[-: ]+', '', text)
    return text.strip()


def ask_batch(model, tokenizer, batch_questions, max_new_tokens=24):
    model_device = next(model.parameters()).device
    prompts = [PROMPT_TEMPLATE.format(question=q['question']) for q in batch_questions]
    inputs = tokenizer(prompts, return_tensors='pt', padding=True, truncation=True).to(model_device)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )
    decoded = tokenizer.batch_decode(outputs, skip_special_tokens=True)
    rows = []
    for q, text in zip(batch_questions, decoded):
        rows.append({'id': q['id'], 'raw_output': text, 'model_answer': parse_answer(text)})
    return rows


def run_eval(model, tokenizer, questions, batch_size=1, max_new_tokens=24):
    predictions = []
    for start in range(0, len(questions), batch_size):
        batch = questions[start:start+batch_size]
        predictions.extend(ask_batch(model, tokenizer, batch, max_new_tokens=max_new_tokens))
    rows = grade_predictions(questions, predictions)
    summary = summarize_rows(rows)
    return predictions, rows, summary


def iter_linear_modules(model):
    for name, module in model.named_modules():
        if isinstance(module, nn.Linear):
            yield name, module


def choose_pilot_layers(model, name_filters=None, limit=8):
    chosen = []
    for name, module in iter_linear_modules(model):
        if name_filters and not any(f in name for f in name_filters):
            continue
        if 'lm_head' in name:
            continue
        chosen.append((name, module))
        if len(chosen) >= limit:
            break
    return chosen


def ternary_snap_(weight: torch.Tensor):
    scale = weight.detach().abs().mean().clamp(min=1e-8)
    q = torch.clamp(torch.round(weight / scale), -1, 1)
    weight.copy_(q * scale)
    return float(scale)
```

```python
tokenizer = AutoTokenizer.from_pretrained(PRUNED_CHECKPOINT, token=HF_TOKEN)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

train_examples = []
for q in questions:
    prompt = PROMPT_TEMPLATE.format(question=q['question'])
    train_examples.append({'text': prompt + ' ' + q['expected_answer']})

dataset = Dataset.from_list(train_examples)


def tokenize_row(row):
    enc = tokenizer(row['text'], truncation=True, max_length=128, padding='max_length')
    enc['labels'] = enc['input_ids'].copy()
    return enc

train_ds = dataset.map(tokenize_row, remove_columns=dataset.column_names)


class RecoveryTrainer(Trainer):
    def __init__(self, teacher_model=None, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.teacher_model = teacher_model

    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        student_outputs = model(**inputs)
        student_logits = student_outputs.logits
        with torch.no_grad():
            teacher_outputs = self.teacher_model(
                input_ids=inputs['input_ids'],
                attention_mask=inputs.get('attention_mask')
            )
            teacher_logits = teacher_outputs.logits
        ce_loss = student_outputs.loss
        kl_loss = torch.nn.functional.kl_div(
            torch.nn.functional.log_softmax(student_logits, dim=-1),
            torch.nn.functional.softmax(teacher_logits, dim=-1),
            reduction='batchmean'
        )
        loss = 0.5 * ce_loss + 0.5 * kl_loss
        return (loss, student_outputs) if return_outputs else loss
```

```python
all_stage_results = []

for target_limit in TARGET_LAYER_LIMITS:
    print(f'\n=== Stage target_limit={target_limit} ===')
    cleanup_memory()

    teacher_model = AutoModelForCausalLM.from_pretrained(
        PRUNED_CHECKPOINT,
        dtype='auto',
        device_map='auto',
        token=HF_TOKEN,
    )
    teacher_model.eval()

    student_model = AutoModelForCausalLM.from_pretrained(
        PRUNED_CHECKPOINT,
        dtype='auto',
        device_map='auto',
        token=HF_TOKEN,
    )
    student_model.eval()

    pilot_layers = choose_pilot_layers(student_model, PILOT_LAYER_NAME_FILTERS, target_limit)
    pilot_layer_names = [name for name, _ in pilot_layers]
    for name, module in student_model.named_modules():
        if name in pilot_layer_names and isinstance(module, nn.Linear):
            ternary_snap_(module.weight.data)

    _, snapped_rows, snapped_summary = run_eval(
        student_model,
        tokenizer,
        questions,
        batch_size=BATCH_SIZE,
        max_new_tokens=MAX_NEW_TOKENS,
    )
    pre_path = timestamped_result_path(RESULTS_DIR, f'07_ternary_expansion_pre_recovery_l{target_limit}')
    save_json({
        'source_model': str(PRUNED_CHECKPOINT),
        'target_layer_limit': target_limit,
        'pilot_layers': pilot_layer_names,
        'summary': snapped_summary,
        'rows': snapped_rows,
    }, pre_path)
    print('Pre-recovery eval salvo em:', pre_path)
    print(snapped_summary)

    lora_config = LoraConfig(
        r=LORA_R,
        lora_alpha=LORA_ALPHA,
        target_modules=['q_proj', 'k_proj', 'v_proj', 'o_proj', 'up_proj', 'down_proj', 'gate_proj'],
        lora_dropout=LORA_DROPOUT,
        bias='none',
        task_type='CAUSAL_LM',
    )
    student_model = get_peft_model(student_model, lora_config)

    recovery_output = CHECKPOINTS_DIR / f'student_ternary_recovered_l{target_limit}'
    args = TrainingArguments(
        output_dir=str(recovery_output),
        per_device_train_batch_size=1,
        gradient_accumulation_steps=8,
        learning_rate=LEARNING_RATE,
        max_steps=TRAIN_STEPS,
        logging_steps=10,
        save_steps=TRAIN_STEPS,
        bf16=torch.cuda.is_available(),
        report_to=[],
        remove_unused_columns=False,
    )

    trainer = RecoveryTrainer(
        model=student_model,
        teacher_model=teacher_model,
        args=args,
        train_dataset=train_ds,
        data_collator=DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False),
    )
    trainer.train()
    trainer.save_model(str(recovery_output))
    tokenizer.save_pretrained(str(recovery_output))
    print('Recovered adapter salvo em:', recovery_output)

    for _name in ['trainer', 'teacher_model', 'student_model']:
        if _name in globals():
            try:
                del globals()[_name]
            except Exception:
                pass
    cleanup_memory()

    peft_model = AutoPeftModelForCausalLM.from_pretrained(
        str(recovery_output),
        token=HF_TOKEN,
        device_map='auto',
        dtype=torch.float16 if torch.cuda.is_available() else None,
        low_cpu_mem_usage=True,
    )
    peft_model.eval()

    _, recovered_rows, recovered_summary = run_eval(
        peft_model,
        tokenizer,
        questions,
        batch_size=BATCH_SIZE,
        max_new_tokens=MAX_NEW_TOKENS,
    )
    post_path = timestamped_result_path(RESULTS_DIR, f'07_ternary_expansion_post_recovery_l{target_limit}')
    save_json({
        'source_model': str(PRUNED_CHECKPOINT),
        'recovered_model': str(recovery_output),
        'target_layer_limit': target_limit,
        'pilot_layers': pilot_layer_names,
        'summary': recovered_summary,
        'rows': recovered_rows,
    }, post_path)
    print('Post-recovery eval salvo em:', post_path)
    print(recovered_summary)

    stage_result = {
        'target_layer_limit': target_limit,
        'pilot_layers': pilot_layer_names,
        'pre_recovery_accuracy': snapped_summary['accuracy'],
        'post_recovery_accuracy': recovered_summary['accuracy'],
        'delta_vs_pre': recovered_summary['accuracy'] - snapped_summary['accuracy'],
        'delta_vs_reference_recovery': recovered_summary['accuracy'] - RECOVERY_BASELINE,
        'pre_result_path': str(pre_path),
        'post_result_path': str(post_path),
        'adapter_path': str(recovery_output),
    }
    all_stage_results.append(stage_result)
    print(stage_result)
```

```python
summary_rows = sorted(all_stage_results, key=lambda r: (r['post_recovery_accuracy'], r['target_layer_limit']), reverse=True)
summary = {
    'reference_recovery_accuracy': RECOVERY_BASELINE,
    'target_accuracy': TARGET_ACCURACY,
    'minimum_acceptable_accuracy': MIN_ACCEPTABLE_ACCURACY,
    'stages': summary_rows,
}
summary_path = timestamped_result_path(RESULTS_DIR, '07_ternary_expansion_summary')
save_json(summary, summary_path)
print('Expansion summary salvo em:', summary_path)
print(json.dumps(summary, ensure_ascii=False, indent=2))

if summary_rows:
    best = summary_rows[0]
    acc = best['post_recovery_accuracy']
    if acc >= TARGET_ACCURACY:
        print('Recomendação: avançar para o próximo salto.')
    elif acc >= MIN_ACCEPTABLE_ACCURACY:
        print('Recomendação: não avançar ainda; testar intermediário ou fortalecer recovery.')
    else:
        print('Recomendação: reduzir o número de camadas e revisar a técnica de recovery.')
```
