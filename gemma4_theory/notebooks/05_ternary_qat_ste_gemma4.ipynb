```python
!pip -q install peft bitsandbytes
import os
import gc
import json
import math
import re
from pathlib import Path
import torch
import torch.nn as nn
from huggingface_hub import login
from transformers import AutoTokenizer, AutoModelForCausalLM

BASE_MODEL_NAME = 'google/gemma-2-2b-it'
PRUNED_CHECKPOINT = Path(CHECKPOINTS_DIR / 'student_pruned')
TERNARY_OUTPUT = Path(CHECKPOINTS_DIR / 'student_ternary_pilot')
PROMPT_TEMPLATE = 'Responda de forma curta e direta.\nPergunta: {question}\nResposta:'
BATCH_SIZE = 1
MAX_NEW_TOKENS = 24
ENABLE_TERNARY_PILOT = False
PILOT_LAYER_LIMIT = 8
PILOT_LAYER_NAME_FILTERS = ['mlp']
PILOT_MODE = 'eval_only'  # eval_only ou checkpoint_only
PROTECT_PATTERNS = ['embed_tokens', 'lm_head', 'norm']
SAMPLE_PER_TENSOR = 20000

HF_TOKEN = os.environ.get('HF_TOKEN')
if IN_COLAB and not HF_TOKEN:
    try:
        from google.colab import userdata
        HF_TOKEN = userdata.get('HF_TOKEN')
    except Exception:
        HF_TOKEN = None

if BASE_MODEL_NAME.startswith('google/gemma') and not HF_TOKEN:
    raise RuntimeError(
        'Gemma no Hugging Face é gated. Crie um secret HF_TOKEN no Colab e aceite a licença do modelo no Hugging Face.'
    )

if HF_TOKEN:
    login(token=HF_TOKEN, add_to_git_credential=False)


def cleanup_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


def checkpoint_has_weights(path: Path) -> bool:
    if not path.exists() or not path.is_dir():
        return False
    return any(path.glob('*.safetensors')) or any(path.glob('*.bin')) or any(path.glob('*.pt'))


def iter_prunable_params(model):
    for name, param in model.named_parameters():
        if param.dim() < 2 or any(p in name for p in PROTECT_PATTERNS):
            continue
        if getattr(param.data, 'is_meta', False):
            continue
        yield name, param


def estimate_global_threshold_from_current_nonzero(model, amount, sample_per_tensor=SAMPLE_PER_TENSOR):
    samples = []
    for _, param in iter_prunable_params(model):
        flat = param.data.detach().abs().reshape(-1)
        nonzero = flat[flat > 0]
        if nonzero.numel() == 0:
            continue
        if nonzero.numel() > sample_per_tensor:
            idx = torch.randperm(nonzero.numel(), device=nonzero.device)[:sample_per_tensor]
            sampled = nonzero[idx]
        else:
            sampled = nonzero
        samples.append(sampled.float().cpu())
    if not samples:
        raise RuntimeError('Nenhum peso não-zero disponível para reconstrução do checkpoint podado.')
    all_scores = torch.cat(samples)
    k = max(1, int(all_scores.numel() * amount))
    threshold = torch.topk(all_scores, k, largest=False).values.max().item()
    return threshold


def apply_incremental_pruning(model, additional_amount):
    threshold = estimate_global_threshold_from_current_nonzero(model, additional_amount)
    for _, param in iter_prunable_params(model):
        data = param.data
        alive = data != 0
        prune_mask = alive & (data.abs() <= threshold)
        data[prune_mask] = 0
    return threshold


def parse_answer(full_text):
    text = full_text.split('Resposta:')[-1].strip()
    text = re.sub(r'^[-: ]+', '', text)
    return text.strip()


def ask_batch(model, tokenizer, batch_questions, max_new_tokens=24):
    prompts = [PROMPT_TEMPLATE.format(question=q['question']) for q in batch_questions]
    inputs = tokenizer(prompts, return_tensors='pt', padding=True, truncation=True).to(model.device)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
        )
    decoded = tokenizer.batch_decode(outputs, skip_special_tokens=True)
    rows = []
    for q, text in zip(batch_questions, decoded):
        rows.append({'id': q['id'], 'raw_output': text, 'model_answer': parse_answer(text)})
    return rows


def run_eval(model, tokenizer, questions, batch_size=1, max_new_tokens=24):
    predictions = []
    for start in range(0, len(questions), batch_size):
        batch = questions[start:start+batch_size]
        predictions.extend(ask_batch(model, tokenizer, batch, max_new_tokens=max_new_tokens))
    rows = grade_predictions(questions, predictions)
    summary = summarize_rows(rows)
    return predictions, rows, summary


def rebuild_pruned_checkpoint_if_needed():
    if checkpoint_has_weights(PRUNED_CHECKPOINT):
        print('Checkpoint podado válido encontrado em', PRUNED_CHECKPOINT)
        return str(PRUNED_CHECKPOINT)

    print('Checkpoint podado ausente/inválido. Reconstruindo para a etapa ternária...')
    latest_rebuilt = sorted(RESULTS_DIR.glob('03_pruned_checkpoint_eval_*.json'))
    latest_pruning = sorted(RESULTS_DIR.glob('02_progressive_pruning_*.json'))
    if not latest_pruning:
        raise FileNotFoundError('Nenhum resultado do 02 encontrado para reconstruir checkpoint podado.')

    selected_target = None
    if latest_rebuilt:
        rebuilt_payload = json.loads(latest_rebuilt[-1].read_text(encoding='utf-8'))
        selected_target = rebuilt_payload.get('selected_target_pruning')

    pruning_payload = json.loads(latest_pruning[-1].read_text(encoding='utf-8'))
    if selected_target is None:
        runs = pruning_payload['runs']
        ranking = sorted(runs[1:], key=lambda r: (-r['summary']['accuracy'], -r.get('global_sparsity', 0.0), r['target_pruning']))
        selected_target = ranking[0]['target_pruning']

    tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL_NAME, token=HF_TOKEN)
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
    model = AutoModelForCausalLM.from_pretrained(BASE_MODEL_NAME, dtype='auto', device_map='auto', token=HF_TOKEN)
    model.eval()

    previous_target = 0.0
    for target in pruning_payload['config']['levels']:
        if target > selected_target + 1e-9:
            break
        additional = max(0.0, target - previous_target)
        if additional > 0:
            threshold = apply_incremental_pruning(model, additional)
            print(f'Reconstrução ternária: target={target:.2%} threshold={threshold:.6f}')
        previous_target = target

    PRUNED_CHECKPOINT.mkdir(parents=True, exist_ok=True)
    model.save_pretrained(PRUNED_CHECKPOINT)
    tokenizer.save_pretrained(PRUNED_CHECKPOINT)
    cleanup_memory()
    print('Checkpoint podado reconstruído em', PRUNED_CHECKPOINT)
    return str(PRUNED_CHECKPOINT)
```

```python
def iter_linear_modules(model):
    for name, module in model.named_modules():
        if isinstance(module, nn.Linear):
            yield name, module


def choose_pilot_layers(model, name_filters=None, limit=8):
    chosen = []
    for name, module in iter_linear_modules(model):
        if name_filters and not any(f in name for f in name_filters):
            continue
        if any(block in name for block in ['lm_head']):
            continue
        chosen.append((name, module))
        if len(chosen) >= limit:
            break
    return chosen


def ternary_snap_(weight: torch.Tensor):
    scale = weight.detach().abs().mean().clamp(min=1e-8)
    q = torch.clamp(torch.round(weight / scale), -1, 1)
    weight.copy_(q * scale)
    return float(scale)


def measure_ternary_ratio(model, layer_names):
    total = 0
    ternary_like = 0
    for name, module in iter_linear_modules(model):
        if name not in layer_names:
            continue
        w = module.weight.detach().float()
        scale = w.abs().mean().clamp(min=1e-8)
        q = torch.clamp(torch.round(w / scale), -1, 1) * scale
        total += w.numel()
        ternary_like += (torch.isclose(w, q, atol=1e-4)).sum().item()
    return ternary_like / total if total else 0.0

pilot_layers = choose_pilot_layers(source_model, PILOT_LAYER_NAME_FILTERS, PILOT_LAYER_LIMIT)
pilot_layer_names = [name for name, _ in pilot_layers]
readiness = {
    'best_variant_from_04': BEST_VARIANT,
    'best_accuracy_from_04': BEST_ACCURACY,
    'source_model': SOURCE_MODEL,
    'source_accuracy': source_summary['accuracy'],
    'pilot_layer_limit': PILOT_LAYER_LIMIT,
    'pilot_layer_name_filters': PILOT_LAYER_NAME_FILTERS,
    'pilot_layers': pilot_layer_names,
    'pilot_layer_count': len(pilot_layer_names),
    'ternary_ratio_before': measure_ternary_ratio(source_model, pilot_layer_names),
}
out_path = timestamped_result_path(RESULTS_DIR, '05_ternary_readiness')
save_json(readiness, out_path)
print('Readiness salvo em:', out_path)
readiness
```

```python
if not ENABLE_TERNARY_PILOT:
    print('Piloto ternário desativado. Isso é o default seguro.')
else:
    pilot_model = AutoModelForCausalLM.from_pretrained(SOURCE_MODEL, dtype='auto', device_map='auto', token=HF_TOKEN)
    pilot_model.eval()
    applied = []
    for name, module in pilot_model.named_modules():
        if name in pilot_layer_names and isinstance(module, nn.Linear):
            scale = ternary_snap_(module.weight.data)
            applied.append({'name': name, 'scale': scale})
    ternary_ratio_after = measure_ternary_ratio(pilot_model, pilot_layer_names)
    _, pilot_rows, pilot_summary = run_eval(pilot_model, source_tokenizer, questions, batch_size=BATCH_SIZE, max_new_tokens=MAX_NEW_TOKENS)
    pilot_result = {
        'source_model': SOURCE_MODEL,
        'pilot_mode': PILOT_MODE,
        'pilot_layers': pilot_layer_names,
        'applied_layers': applied,
        'ternary_ratio_after': ternary_ratio_after,
        'summary': pilot_summary,
        'rows': pilot_rows,
    }
    out_path = timestamped_result_path(RESULTS_DIR, '05_ternary_pilot_eval')
    save_json(pilot_result, out_path)
    print('Pilot eval salvo em:', out_path)
    print(pilot_summary)
    if PILOT_MODE == 'checkpoint_only':
        TERNARY_OUTPUT.mkdir(parents=True, exist_ok=True)
        pilot_model.save_pretrained(TERNARY_OUTPUT)
        source_tokenizer.save_pretrained(TERNARY_OUTPUT)
        print('Checkpoint ternário piloto salvo em:', TERNARY_OUTPUT)
```
